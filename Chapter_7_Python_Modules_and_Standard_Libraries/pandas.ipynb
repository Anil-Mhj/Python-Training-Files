```python
import pandas as pd

data = [
    {"name": "John", "age": 25, "city": "New York"},
    {"name": "Alice", "age": 32, "city": "San Francisco"},
    {"name": "Nate", "age": 41, "city": "Seattle"},
]

data.extend([{"name": "Emily", "age": 29, "city": "Boston"}])

df = pd.DataFrame(data)

df
```

```python
additional_data=[
    {'name':'Nate','age':29,'city':'Austin','new_value':58},
    {'name':'Peter','age':27,'city':'Boston','new_value':54},
    {'name':'Marshall','age':24,'city':'Austin','new_value':48},
    {'name':'Mark','age':28,'city':'New York','new_value':56},
    {'name':'Diana','age':43,'city':'Dallas','new_value':86},
    {'name':'Chris','age':39,'city':'Chicago','new_value':66},
    {'name':'Brown','age':32,'city':'Portland','new_value':64},
    {'name':'Steve','age':38,'city':'Austin','new_value':68},
    {'name':'Whitney','age':37,'city':'Portland','new_value':74},
    {'name':'John','age':25,'city':'New York','new_value':50},
    {'name':'Alice','age':33,'city':'San Francisco','new_value':66},
    {'name': 'to be removed', "age": 44, 'city':"Portland",'new_value':None}
]

new_records=pd.DataFrame(additional_data)
df=pd.concat([df,new_records],ignore_index=True)
df
```

```python
df.dropna(inplace=True)
```

```python
df=df.rename(columns={'new_value':'salary'})
df['salary']=df['salary'].astype(str)+'k'
```

```python
df
```

```python
new_data=[
    {'name':"Sarah",'age':None,'city':"Houston",'salary':'50.0k'},
    {'name':"Jessy",'age':None,'city':"New York",'salary':'68.0k'},
    {'name':"Cody",'age':None,'city':"Chicago",'salary':'55.0k'}
]
df=pd.concat([df,pd.DataFrame(new_data)],ignore_index=True)
df
```

```python
boston_filtered=df[
    (df['city']=='Boston') & (df['age'].between(25,30)) & (df['salary'].str.replace('k','').astype(float)>34)
]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 5))

plt.subplot(131)
df['city'].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Distribution by City")

plt.subplot(132)
avg_salary = df.groupby("city")["salary"].apply(
    lambda x: x.str.replace("k", "").astype(float).mean()
)
avg_salary.plot(kind="bar")
plt.title("Average Salary by City")
plt.xticks(rotation=45)
plt.ylabel("Average Salary (k)")

plt.subplot(133)
salary_numeric = df["salary"].str.replace("k", "").astype(float)
plt.scatter(df["age"], salary_numeric)
plt.xlabel("Age")
plt.ylabel("Salary (k)")
plt.title("Age vs Salary")

plt.tight_layout()
plt.show()
```

```python
import numpy as np

import matplotlib.pyplot as plt

x = [0, 1, 2, 3, 4]
y = [3, 2, 6, 8, 10]

plt.figure(figsize=(6, 4), dpi=88)

#Plot first line with points and solid line
plt.plot(x, y, "r.-", label="2x")

x2 = np.arange(0, 4.5, 0.5)

plt.plot(x2[4:], x2[4:] ** 3, "b*--", label="x^3")
plt.plot(x2[0:5], x2[0:5] ** 3, "g.-", label="x^3")

plt.title("My first graph!", fontsize=28)
# plt.rcParams["font.family"] = "monospace"
# plt.rcParams["font.monospace"] = ["FreeMono"]

plt.xlabel("x axis")
plt.ylabel("y axis")

plt.xticks([0, 1, 2, 3, 4])

plt.legend()

plt.savefig("mygraph.png", dpi=280)

plt.show()
```

```python
labels = ["A", "B", "C"]
values = [1, 4, 3]
plt.title("My first Bar Charts!", fontdict={"fontsize": 20})

bars = plt.bar(labels, values)
plt.figure(figsize=(6, 4))
# bars[0].set_hatch("+")
# bars[1].set_hatch("o")
# bars[2].set_hatch("*")

patterns = ["+", "o", "*"]
for bar in bars:
    bar.set_hatch(patterns.pop(0))

plt.savefig("mycharts.png", dpi=280)

plt.show()
```
