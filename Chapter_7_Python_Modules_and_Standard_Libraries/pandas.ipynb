```python
import pandas as pd

data = [
    {"name": "John", "age": 25, "city": "New York"},
    {"name": "Alice", "age": 32, "city": "San Francisco"},
    {"name": "Nate", "age": 41, "city": "Seattle"},
]

data.extend([{"name": "Emily", "age": 29, "city": "Boston"}])

df = pd.DataFrame(data)

df
```
